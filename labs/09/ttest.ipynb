```python
# 1. Dataset selection 

from ucimlrepo import fetch_ucirepo 

# Code used from the link shown below
# Link adult dataset - https://archive.ics.uci.edu/dataset/2/adult

# fetch dataset 
adult = fetch_ucirepo(id=2) 

# data (as pandas dataframes) 
x = adult.data.features 
y = adult.data.targets 

# metadata 
print(adult.metadata) 

# variable information 
print(adult.variables) 
```

#### 2. "Identify suitable samples within the dataset that allow you to test a meaningful hypothesis using a t-test. "   
Choosen variables: `hours-per-week` and `income`

groups: 
- mu1 -> mean weekly working hours for individuals earning >50k
- mu2 -> mean weekly working hours for individuals earning <=50k  

hypothesis (two sided):
- H0 -> mu1 = mu2 -> There is no difference in mean weekly working hours between income groups.
- H1 -> mu1 != mu2 -> There is a difference in mean weekly working hours between income groups.
- alpha = 0.05


#### 3. "Decide which type of t-test is appropriate (e.g., one-sample, independent two-sample, or paired)."  

I use an independent two-sample t-test:  
- two independent groups (>50k and <=50k)
- Every Value represents a person that can't earn <50k and >=50k

```python
# 4. Implement the t-test in Python.
import pandas as pd
from scipy.stats import ttest_ind

# Create the DataFrame
df = pd.concat([x, y], axis=1)

# Drop missing Values
df = df.dropna()

# Short visualization
print(df.groupby('income')['hours-per-week'].agg(['count', 'mean', 'std']))

# Create 2 samples
high_income_hours = df[df["income"] == ">50K"]["hours-per-week"]
low_income_hours = df[df["income"] == "<=50K"]["hours-per-week"]

# Independent 2 side t-test - no equal variance
t_value, p_value = ttest_ind(
    high_income_hours,
    low_income_hours,
    equal_var=False 
)
#t_stat, p_value
print()
print(f"t = {t_value} and p = {p_value}")
```

#### 5. Interpret and explain the results clearly, linking them back to your hypothesis
t ~ 45.12 & p = 0

> Since the p-value is far below the chosen significance level of α = 0.05 (p < 0.001), the null hypothesis (μ₁ = μ₂) is rejected. This indicates a highly statistically significant difference in mean weekly working hours between the two income groups. The positive t-statistic suggests that individuals with higher income work more hours per week on average than those with lower income. (*Sentence corrected by ChatGPT)
